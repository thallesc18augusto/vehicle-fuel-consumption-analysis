```python

#Outlier Detection using the IQR method
outlier_summary = []

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    
    outlier_summary.append({
        'Variable': col,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outliers': len(outliers),
        'Percentage (%)': len(outliers) / len(df) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
#Creating variable for the target and features
target = 'Combined (L/100 km)'

features = [
    'Model year',
    'Make',
    'Vehicle class',
    'Engine size (L)',
    'Cylinders',
    'Transmission',
    'Fuel type'
]

X = df[features]
y = df[target]

print("Features:", X.columns.tolist())
print("Target:", y.name)
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
#Separating categorical and numerical features
categorical_features = [
    'Make',
    'Vehicle class',
    'Transmission',
    'Fuel type'
]

numerical_features = [
    'Model year',
    'Engine size (L)',
    'Cylinders'
]

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
%pip install scikit-learn
```

```python
#Splitting the dataset into training and test sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training set:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTest set:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
#Checking for unseen categories in the test set
for col in categorical_features:
    train_categories = set(X_train[col].unique())
    test_categories = set(X_test[col].unique())
    
    unseen_categories = test_categories - train_categories
    
    print(f"{col}:")
    print("Unseen categories in test set:", unseen_categories)
    print()
```

```python
#Creating a preprocessing pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ('num', 'passthrough', numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
#Transforming the training set
X_train_transformed = preprocessor.fit_transform(X_train)

print("Original number of features:", X_train.shape[1])
print("Transformed number of features:", X_train_transformed.shape[1])
```

```python
#Transforming the test set
X_test_transformed = preprocessor.transform(X_test)

print("Training transformed shape:", X_train_transformed.shape)
print("Test transformed shape:", X_test_transformed.shape)
```

```python
#training a Linear Regression model
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(X_train_transformed, y_train)

y_pred_linear = linear_model.predict(X_test_transformed)

print("Linear Regression model trained successfully.")
```

```python
#Evaluating the Linear Regression model
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_linear = mean_absolute_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))
r2_linear = r2_score(y_test, y_pred_linear)

print("Linear Regression Performance")
print(f"MAE:  {mae_linear:.4f}")
print(f"RMSE: {rmse_linear:.4f}")
print(f"R²:   {r2_linear:.4f}")
```

```python
# Random Forest model
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_transformed, y_train)

y_pred_rf = rf_model.predict(X_test_transformed)

print("Random Forest model trained successfully.")
```

```python
#Evaluating the Random Forest model
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Performance")
print(f"MAE:  {mae_rf:.4f}")
print(f"RMSE: {rmse_rf:.4f}")
print(f"R²:   {r2_rf:.4f}")
```

```python
model_results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest'],
    'MAE': [mae_linear, mae_rf],
    'RMSE': [rmse_linear, rmse_rf],
    'R2': [r2_linear, r2_rf]
})

model_results
```

```python
#tuning the Random Forest model using RandomizedSearchCV
from sklearn.model_selection import RandomizedSearchCV

param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

rf_tuning = RandomizedSearchCV(
    estimator=RandomForestRegressor(
        random_state=42,
        n_jobs=-1
    ),
    param_distributions=param_grid,
    n_iter=20,
    cv=5,
    scoring='neg_root_mean_squared_error',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_tuning.fit(X_train_transformed, y_train)

print("Best parameters:")
print(rf_tuning.best_params_)

print("\nBest CV RMSE:")
print(-rf_tuning.best_score_)
```

```python
# Selecting the best Random Forest model
best_rf_model = rf_tuning.best_estimator_

y_pred_rf_tuned = best_rf_model.predict(X_test_transformed)

mae_rf_tuned = mean_absolute_error(y_test, y_pred_rf_tuned)
rmse_rf_tuned = np.sqrt(mean_squared_error(y_test, y_pred_rf_tuned))
r2_rf_tuned = r2_score(y_test, y_pred_rf_tuned)

print("Tuned Random Forest Performance")
print(f"MAE:  {mae_rf_tuned:.4f}")
print(f"RMSE: {rmse_rf_tuned:.4f}")
print(f"R²:   {r2_rf_tuned:.4f}")
```

```python
#Comparing the performance of the models
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred_rf_tuned, alpha=0.5)

min_value = min(y_test.min(), y_pred_rf_tuned.min())
max_value = max(y_test.max(), y_pred_rf_tuned.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle='--'
)

plt.xlabel('Actual Combined Fuel Consumption (L/100 km)')
plt.ylabel('Predicted Combined Fuel Consumption (L/100 km)')
plt.title('Actual vs. Predicted Fuel Consumption - Tuned Random Forest')

plt.tight_layout()
plt.show()
```

```python
#Residual Analysis
residuals = y_test - y_pred_rf_tuned

plt.figure(figsize=(8, 6))

plt.scatter(y_pred_rf_tuned, residuals, alpha=0.5)
plt.axhline(y=0, linestyle='--')

plt.xlabel('Predicted Combined Fuel Consumption (L/100 km)')
plt.ylabel('Residuals')
plt.title('Residual Analysis - Tuned Random Forest')

plt.tight_layout()
plt.show()
```

```python
#Classification model to define the median
```

```python
classification_threshold = y_train.median()

print("Classification threshold:", classification_threshold)
```

```python
#training a classification model using the median as the threshold
y_train_class = (y_train > classification_threshold).astype(int)
y_test_class = (y_test > classification_threshold).astype(int)

print("Training class distribution:")
print(y_train_class.value_counts())

print("\nTest class distribution:")
print(y_test_class.value_counts())
```

```python
#Preparing the data for classification
X_train_class = X_train.copy()
X_test_class = X_test.copy()

print("Classification training features:", X_train_class.shape)
print("Classification test features:", X_test_class.shape)
```

```python
# Creating a preprocessing pipeline for classification
X_train_class_transformed = preprocessor.transform(X_train_class)
X_test_class_transformed = preprocessor.transform(X_test_class)

print("Classification training shape:", X_train_class_transformed.shape)
print("Classification test shape:", X_test_class_transformed.shape)
```

```python
#logistic regression model for classification
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=3000,
    random_state=42
)

logistic_model.fit(
    X_train_class_transformed,
    y_train_class
)

y_pred_logistic = logistic_model.predict(
    X_test_class_transformed
)

y_prob_logistic = logistic_model.predict_proba(
    X_test_class_transformed
)[:, 1]

print("Logistic Regression trained successfully!")
```

```python
#Evaluating the Logistic Regression model
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy_logistic = accuracy_score(y_test_class, y_pred_logistic)
precision_logistic = precision_score(y_test_class, y_pred_logistic)
recall_logistic = recall_score(y_test_class, y_pred_logistic)
f1_logistic = f1_score(y_test_class, y_pred_logistic)
roc_auc_logistic = roc_auc_score(y_test_class, y_prob_logistic)

print("Logistic Regression Performance:")
print(f"Accuracy:  {accuracy_logistic:.4f}")
print(f"Precision: {precision_logistic:.4f}")
print(f"Recall:    {recall_logistic:.4f}")
print(f"F1-Score:  {f1_logistic:.4f}")
print(f"ROC-AUC:   {roc_auc_logistic:.4f}")
```

```python
#plotting the confusion matrix for the Logistic Regression model
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm_logistic = confusion_matrix(y_test_class, y_pred_logistic)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Efficient", "Less Efficient"]
)

disp.plot()
plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
#Random Forest training for classification
from sklearn.ensemble import RandomForestClassifier

rf_classifier = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_classifier.fit(
    X_train_class_transformed,
    y_train_class
)

y_pred_rf_class = rf_classifier.predict(
    X_test_class_transformed
)

y_prob_rf_class = rf_classifier.predict_proba(
    X_test_class_transformed
)[:, 1]

print("Random Forest Classifier trained successfully!")
```

```python

#Evaluating the Random Forest Classifier
accuracy_rf_class = accuracy_score(y_test_class, y_pred_rf_class)
precision_rf_class = precision_score(y_test_class, y_pred_rf_class)
recall_rf_class = recall_score(y_test_class, y_pred_rf_class)
f1_rf_class = f1_score(y_test_class, y_pred_rf_class)
roc_auc_rf_class = roc_auc_score(y_test_class, y_prob_rf_class)

print("Random Forest Classifier Performance:")
print(f"Accuracy:  {accuracy_rf_class:.4f}")
print(f"Precision: {precision_rf_class:.4f}")
print(f"Recall:    {recall_rf_class:.4f}")
print(f"F1-Score:  {f1_rf_class:.4f}")
print(f"ROC-AUC:   {roc_auc_rf_class:.4f}")
```

```python
#TUNING THE RANDOM FOREST CLASSIFIER USING RANDOMIZEDSEARCHCV
from sklearn.model_selection import RandomizedSearchCV

classifier_param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': ['sqrt', 'log2', None]
}

rf_classifier_tuning = RandomizedSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),
    param_distributions=classifier_param_grid,
    n_iter=20,
    cv=5,
    scoring='roc_auc',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_classifier_tuning.fit(
    X_train_class_transformed,
    y_train_class
)

print("Best parameters:")
print(rf_classifier_tuning.best_params_)

print("\nBest CV ROC-AUC:")
print(rf_classifier_tuning.best_score_)
```

```python
#Optimize model
best_rf_classifier = rf_classifier_tuning.best_estimator_

y_pred_rf_class_tuned = best_rf_classifier.predict(
    X_test_class_transformed
)

y_prob_rf_class_tuned = best_rf_classifier.predict_proba(
    X_test_class_transformed
)[:, 1]

accuracy_rf_class_tuned = accuracy_score(
    y_test_class,
    y_pred_rf_class_tuned
)

precision_rf_class_tuned = precision_score(
    y_test_class,
    y_pred_rf_class_tuned
)

recall_rf_class_tuned = recall_score(
    y_test_class,
    y_pred_rf_class_tuned
)

f1_rf_class_tuned = f1_score(
    y_test_class,
    y_pred_rf_class_tuned
)

roc_auc_rf_class_tuned = roc_auc_score(
    y_test_class,
    y_prob_rf_class_tuned
)

print("Tuned Random Forest Classifier Performance:")
print(f"Accuracy:  {accuracy_rf_class_tuned:.4f}")
print(f"Precision: {precision_rf_class_tuned:.4f}")
print(f"Recall:    {recall_rf_class_tuned:.4f}")
print(f"F1-Score:  {f1_rf_class_tuned:.4f}")
print(f"ROC-AUC:   {roc_auc_rf_class_tuned:.4f}")
```

```python
#Confusion matrix final model
cm_rf_tuned = confusion_matrix(
    y_test_class,
    y_pred_rf_class_tuned
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_rf_tuned,
    display_labels=["Efficient", "Less Efficient"]
)

disp.plot()
plt.title("Tuned Random Forest - Confusion Matrix")
plt.show()
```

```python
#roc curve for the classification models
from sklearn.metrics import roc_curve

fpr_logistic, tpr_logistic, _ = roc_curve(
    y_test_class,
    y_prob_logistic
)

fpr_rf, tpr_rf, _ = roc_curve(
    y_test_class,
    y_prob_rf_class_tuned
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr_logistic,
    tpr_logistic,
    label=f"Logistic Regression (AUC = {roc_auc_logistic:.4f})"
)

plt.plot(
    fpr_rf,
    tpr_rf,
    label=f"Tuned Random Forest (AUC = {roc_auc_rf_class_tuned:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle='--',
    label="Random Classifier"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Classification Models")
plt.legend()
plt.show()
```
